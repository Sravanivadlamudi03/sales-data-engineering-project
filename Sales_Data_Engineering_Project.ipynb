```python
import pandas as pd

customers = pd.DataFrame({
    "customer_id":[101,102,103,104,105],
    "customer_name":["John Smith","Sarah Johnson","Michael Brown","Emily Davis","David Wilson"],
    "city":["Dallas","Austin","Phoenix","Chicago","Seattle"],
    "state":["TX","TX","AZ","IL","WA"]
})

customers
```

```python
customers = pd.DataFrame({
    "customer_id": [101, 102, 103, 104, 105],
    "customer_name": ["John Smith", "Sarah Johnson", "Michael Brown", "Emily Davis", "David Wilson"],
    "city": ["Dallas", "Austin", "Phoenix", "Chicago", "Seattle"],
    "state": ["TX", "TX", "AZ", "IL", "WA"]
})

customers
```

```python
products = pd.DataFrame({
    "product_id": [201, 202, 203, 204, 205],
    "product_name": ["Laptop", "Phone", "Headphones", "Monitor", "Keyboard"],
    "category": ["Electronics", "Electronics", "Accessories", "Electronics", "Accessories"],
    "price": [1200, 800, 150, 300, 80]
})

products
```

```python
sales = pd.DataFrame({
    "sale_id": [1, 2, 3, 4, 5],
    "customer_id": [101, 102, 103, 104, 105],
    "product_id": [201, 202, 203, 204, 205],
    "sale_date": ["2024-01-01", "2024-01-02", "2024-01-03", "2024-01-04", "2024-01-05"],
    "quantity": [1, 2, 3, 1, 4],
    "amount": [1200, 1600, 450, 300, 320],
    "region": ["South", "South", "West", "Midwest", "West"]
})

sales
```

```python
sales_customer = sales.merge(customers, on="customer_id", how="left")

final_sales_data = sales_customer.merge(products, on="product_id", how="left")

final_sales_data
```

```python
sales_by_region = final_sales_data.groupby("region")["amount"].sum().reset_index()

sales_by_region
```

```python
sales_by_product = final_sales_data.groupby("product_name")["amount"].sum().reset_index()

sales_by_product
```

```python
customers.to_csv("data/customers.csv", index=False)
products.to_csv("data/products.csv", index=False)
sales.to_csv("data/sales.csv", index=False)
final_sales_data.to_csv("data/final_sales_data.csv", index=False)

print("All CSV files saved successfully!")
```

```python
import matplotlib.pyplot as plt

sales_by_region.plot(
    x="region",
    y="amount",
    kind="bar"
)

plt.title("Sales by Region")
plt.show()
```
